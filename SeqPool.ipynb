### Calculating the desired pool concentration in ng/uL

$$ \frac{(DNA\ concentration\ ng/uL)}{660 g/mol} * 10^6 = concentration\ in\ nM$$

$10^6$ is from converting $M = mol/L$ to $nM$ and an additional conversiton from $L$ to $uL$ 

The AVITI sequencer requires at least 10uL of 13nM of DNA. After performing this calculation with this target concentration in mind, I find that my pool needs to be 4.179 ng/uL. To play it on the safe side, I round up to 5 ng/uL in the following code. This is taken into account in the target_ng variable

```python

volumes = np.array([18.2, 23.3, 14.9, 21.3, 4.7, 24.3, 19.0, 12.6, 27.4, 13.7]) #in ng/uL, blank subtracted
total_vol = 50 #in ul 
target_ng = 5 * total_vol #ng
"""
The value of 5 comes from the DNA concentration calculation where bp = 550 

"""

target_each = target_ng / np.size(volumes)

uLs = (target_each / volumes) * 5 #chose this number 5 after noticing my volumes were very low
resultant_vol = np.sum(uLs)
print(uLs)
print(resultant_vol)
```
